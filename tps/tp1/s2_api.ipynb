# APIs: requêtes HTTP

## Imports

```python
import json
import requests
```

## Utiliser Gutendex pour récupérer des informations sur un auteur

https://gutendex.com/

```python
author_name = "Hugo"

url = "https://gutendex.com/books/"
params = {'search': author_name, 'languages': 'fr'}
resp = requests.get(url, params=params)
result = resp.json()
try:
    print(f"Number of books in French: {result['count']}")
    for book in result['results'][:5]:
        author = book['authors'][0]
        print(f"{book['title']} - {author['name']} ({author['birth_year']} - {author['death_year']})")
except (KeyError, IndexError):
    print("Unknown author, please check the name")
```
